```python
marks = [88, 72, 95, 61, 79]

# Access O(1)
print(marks[0])
print(marks[1])
print(marks[2])
print(marks[-1])

# Add to end O(1)
marks.append(100)
print(marks)

# Specific posn O(n)
marks.insert(2,55)
print(marks)

# Delete
marks.pop() # remove the last element
print(marks)

marks.pop(3) # remove from a specific index
print(marks)

marks.remove(61) # remove by a value
print(marks)
```

Operation    Where   Speed  
Acess        any      O(1)  
Insert       end      O(1)
insert      middle    O(n)  
delete       end      O(1)
delete      middle    O(n)
slice     any range   O(K)

Slicing

```python
marks = [88, 72, 95, 61, 79]
print(marks[1:4])
print(marks[::-1])
print(marks[::2])
```
