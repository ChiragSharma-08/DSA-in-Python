Left to right

```python
marks = [88, 72, 95, 61, 79]

for mark in marks:
    print(mark)

for i in range(len(marks)):
    print(marks[i])
```

Right to Left 

```python
marks = [88, 72, 95, 61, 79]

for mark in reversed(marks):
    print(mark)

for mark in marks[::-1]:
    print(mark)

for i in range(len(marks) -1, -1, -1):
    print(marks[i])
```

```python
classroom = [
    [85, 90, 78], #r1
    [72, 88, 91], #r2
    [95, 60, 83], #r3
]

for row in classroom:
    for mark in row:
        print(mark, end=" ")
```
